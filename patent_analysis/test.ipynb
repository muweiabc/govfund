```python

df = pd.DataFrame({
    "category": ["A", "A", "B", "B", "B", "C"],
    "tt": [10, 20, 5, 15, 25, 30]
})
result = df.groupby("category")['tt'].agg(
    ['count','sum']).reset_index() # 每组 value 总和

print(result)
with pd.ExcelWriter("test.xlsx") as writer:
    result.to_excel(writer, sheet_name='test', index=False)
```
